# 2035 ride-service market sizing
Top-down sizing from a sourced 2026 anchor (S1–S3 in `docs/SOURCES.md`). Segment splits are labelled assumptions in `assumptions.yaml`. The same logic runs live in the app (`web/lib/engine/market.ts`); a pytest keeps them in sync.

```python
import pandas as pd
from market_sizing import load_assumptions, size_market
a = load_assumptions()
base = size_market(a)
print(f"2035 taxi market ≈ US${base['taxiMarket2035UsdBn']:.1f} bn; ride-hailing ≈ ₹{base['rideHailing2035Cr']/1e5:.2f} lakh crore")
pd.DataFrame(base['segments'])[['label','tripsPerDayLakh','avgTripKm','paidKmBn','farePerKm','marginPct','marginCr','pool']].round(2)
```

## Low / base / high cases
Sources disagree on growth (Mordor 17.9% vs TechSci 6.2% CAGR), so we bracket the result.

```python
cases = {'low (TechSci-like 6.2%)': (0.062, 0.05), 'base': (None, None), 'high': (0.20, 0.12)}
rows = []
for name, (c1, c2) in cases.items():
    r = size_market(a, cagr1=c1, cagr2=c2)
    rows.append({'case': name, 'taxi_2035_usd_bn': round(r['taxiMarket2035UsdBn'], 1), 'ride_hailing_lakh_cr': round(r['rideHailing2035Cr']/1e5, 2)})
pd.DataFrame(rows)
```
